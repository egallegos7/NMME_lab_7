### Problem 1: Newton-Cotes Quadrature [0.6 pts]
- Part (a): Implement Midpoint, Trapezoidal, and Simpson's rules [0.3 pts]
- Part (b): Test on $f(x) = x^2 + 1$ over $[0, 2]$ and compute errors [0.3 pts]

```python
# --- Part (a) ---
def newton_cotes(f, a, b):
    """
    Approximates the integral of f from a to b using Newton-Cotes rules.
    Returns a dictionary with keys: 'midpoint', 'trapezoidal', 'simpsons'.
    """

    M = (b - a) * f((a + b) / 2)

    T = (b - a) / 2 * (f(a) + f(b))

    S = (b - a) / 6 * (f(a) + 4*f((a + b) / 2) + f(b))

    return {'midpoint': M, 'trapezoidal': T, 'simpsons': S}

# --- Part (b) ---
f1 = lambda x: x**2 + 1
a1, b1 = 0, 2
exact1 = 14 / 3

results = newton_cotes(f1, a1, b1)

print(f"Exact: {exact1:.4f}")

for method, val in results.items():
    error = abs(val - exact1)
    print(f"{method.capitalize():<13}: {val:.4f} | Error: {error:.4f}")
```

### Problem 2: Gauss-Legendre Quadrature [0.6 pts]
- Part (a): Implement 2-point Gauss-Legendre with coordinate transformation [0.3 pts]
- Part (b): Test on $f(x) = x^3 + 2x$ over $[0, 0.8]$ [0.3 pts]

```python
# --- Part (a) ---
def gauss_legendre_2(f, a, b):
    """
    Approximates the integral of f from a to b using 2-point Gauss-Legendre.
    """
    # Standard GL2 nodes and weights on [-1, 1]
    xd = np.array([-1/np.sqrt(3), 1/np.sqrt(3)])
    w = np.array([1.0, 1.0])

    x = ((b - a) / 2) * xd + ((a + b) / 2)

    integral = (b - a) / 2 * np.sum(w * f(x))

    return integral

# --- Part (b) ---
f2 = lambda x: x**3 + 2*x
a2, b2 = 0, 0.8
exact2 = 0.7424

gl_approx = gauss_legendre_2(f2, a2, b2)

error2 = abs(gl_approx - exact2)

print(f"Exact integral:       {exact2:.4f}")
print(f"GL2 approximation:    {gl_approx:.4f}")
print(f"Absolute error:       {error2:.4f}")
```

### Problem 3: Numerical Differentiation via Vandermonde [0.8 pts]
- Part (a): Construct the scaled Vandermonde matrix and find weights [0.4 pts]
- Part (b): Approximate the derivative of $f(x) = \sin(x)$ at $x=1$ [0.4 pts]

```python
# --- Part (a) ---
def get_diff_weights(delta, max_deriv):
    """
    Computes differentiation weights using a scaled Vandermonde matrix.
    delta: array of node offsets (e.g., [0, 1, 2, 3])
    max_deriv: maximum derivative order to compute
    Returns: C matrix where row k contains weights for the k-th derivative.
    """
    n = len(delta)
    j = np.arange(n)

    j_fact = np.array([math.factorial(k) for k in j])

    V = (delta[:, None] ** j) / j_fact

    C = np.linalg.inv(V)

    return C[:max_deriv+1, :]

# Test the function
delta = np.array([0, 1, 2, 3])

C = get_diff_weights(delta, 3)

print("Differentiation Weight Matrix C (up to 3rd derivative):")

print(C)
```

```python
# --- Part (b) ---
x_eval = 1.0
h = 0.1

f_vals = np.sin(x_eval + delta*h)


df_approx = (1 / h) * np.sum(C[1, :] * f_vals)

df_exact = np.cos(x_eval)

print(f"Approximate f'({x_eval}): {df_approx:.6f}")
print(f"Exact f'({x_eval}):       {df_exact:.6f}")
print(f"Absolute error:           {abs(df_approx - df_exact):.6f}")
```
